```python
import scipy as sc
import numpy as np
import sympy as sp
import matplotlib as mpl
import matplotlib.pyplot as plt
```

The sperical harmonics:
\begin{align*}
  Y^{m}_{l}(\theta,\phi) = \epsilon\sqrt{\frac{(2l +1)(l-|m|)!}{4\pi(l+|m|)!}}e^{im\phi}P^{m}_{l}(\cos{\theta})
\end{align*}

The associated Legendre functions:
\begin{align*}
  P^{m}_{l}(x)=(1-x^{2})^{\frac{|m|}{2}}(\frac{d}{dx})^{|m|}P_{l}(x)
\end{align*}

Defined by the Rodrigues formula:
\begin{align*}
  P_{l}(x)=\frac{1}{2^{l}l!}(\frac{d}{dx})^{l}(x^{2}-1)^{l}
\end{align*}

```python
m = 0
l = 1
the, phi = np.linspace(0, np.pi, 200), np.linspace(0, 2*np.pi, 200)
THE, PHI = np.meshgrid(the, phi)

Yml = sc.special.sph_harm(m, l, THE, PHI)
#plt.plot(the,Yml)
```

```python
fig = plt.figure()
ax = fig.add_subplot(projection='3d')

R = np.real(Yml)
X = R * np.sin(PHI) * np.cos(THE)
Y = R * np.sin(PHI) * np.sin(THE)
Z = R * np.cos(PHI)

ax.plot_surface(X, Y, Z, cmap = plt.cm.YlGnBu_r)

ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
ax.set_aspect('equal')
```
