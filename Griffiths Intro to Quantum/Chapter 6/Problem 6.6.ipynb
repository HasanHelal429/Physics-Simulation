```python
%matplotlib tk
import scipy as sc
import numpy as np
import sympy as sp
import matplotlib as mpl
import matplotlib.pyplot as plt
import scienceplots

plt.style.use(['science','notebook', 'grid'])
```

```python
N = 100
dx = 1 / N

x = np.linspace(0, 1, N)
idx = (x < 0.25) | (x > 0.75) 
V = np.zeros(N)
V[idx] = -10


plt.plot(x, V, label='Potential $V(x)', color='blue', lw=2)
```

```python
qs = np.linspace(-np.pi, np.pi, 100)
Enq = []
psinq = []


laplacian = -2 * np.eye(N) + np.eye(N, k=1) + np.eye(N, k=-1)
laplacian[0,-1] = 1
laplacian[-1,0] = 1
deriv = np.eye(N, k=1) - np.eye(N, k=-1)
deriv[-1,0] = 1
deriv[0,-1] = -1
for q in qs:

    H = -laplacian * 0.25 - 1j * q * deriv + np.diag(V + q**2 / 2)
    En, psin = np.linalg.eigh(H)
    psinq.append(psin)
    Enq.append(En)

```

```python
np.shape(Enq)
E1q = np.array(Enq).T[:][0]
E2q = np.array(Enq).T[:][1]

plt.figure(figsize=(8,6))
plt.plot(qs, E1q, label='E1(q)', color='red', lw=2)
plt.plot(qs, E2q, label='E2(q)', color='blue', lw=2)
plt.xlabel(r'$\frac{q}{a}$')
plt.ylabel(r'$\frac{m}{\hbar^2}E_n(q)$')
plt.legend()
```

```python
np.shape(psinq)
psinq = np.array(psinq)
for k in range(0, 3):
    psi1k = psinq[0, :, k]
    plt.plot(x, psi1k.real**2 + psi1k.imag**2, label='psi1k', lw=2)
```
