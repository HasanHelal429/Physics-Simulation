```python
import numpy as np
import sympy as sp
import matplotlib.pyplot as plt
import scipy as sc
```

```python
def get_A(b):
    x = np.linspace(-10, 10, 1000)
    return 1 / np.sqrt(np.trapz((1/(x**2+b**2))**2,x))
def psi(x, b):
    return  get_A(b) / (x**2 + b**2)

def Potential(x):
    return  x**2/2
```

```python
x = np.linspace(-10, 10, 1000)
V = Potential(x)
b = np.linspace(-1, 1, 100)
V_avg = [np.trapz(V*psi(x, b[0])**2, x)]
for i in range(1, len(b)):
    V_avg = np.append(V_avg, np.trapz(V*psi(x, b[i])**2, x)) 
E_g = np.ones(len(b)) * 0.5 + V_avg
plt.plot(b, E_g, label='Ground State Energy')
```
