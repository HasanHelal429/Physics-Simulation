A comprehensive guide to the harmonic oscillator.

The simplest harmonic oscillator is described by the differential equation:
$$ m\ddot{x} = -kx $$

However a more comprehensive description of a harmonic oscillator has a damping term.
$$ m\ddot{x} = -kx - b\dot{x} $$

```python
%matplotlib tk
import scipy as sc
import numpy as np
import sympy as sp
import matplotlib as mpl
import matplotlib.pyplot as plt
import scienceplots

plt.style.use(['science','notebook', 'grid'])
```

```python
N = 250
t, m, k, b = sp.symbols('t m k b')
x = sp.symbols('x', cls=sp.Function)
x = x(t)

x_d = sp.diff(x, t)
x_dd = sp.diff(x_d, t)
```

```python
D = m*x_dd + k*x + b*x_d
sol = sp.solve(D, x_dd, simplify = False, rational=False)
sol[0]

dvdt_f = sp.lambdify((t, m, k, b, x, x_d), sol[0])
dxdt_f = sp.lambdify(x_d, x_d)
```

```python
def dSdt(S, t, m, k, b):
    x, x_d = S
    return [
        dxdt_f(x_d),
        dvdt_f(t, m, k, b, x, x_d)
    ]
```

```python
t = np.linspace(0, 40, N)
m = 1
k = 1
b = 0.3
ans = sc.integrate.odeint(dSdt, y0=[1,0], t=t, args=(m, k, b))
```

```python
plt.plot(t, ans.T[0])
```
