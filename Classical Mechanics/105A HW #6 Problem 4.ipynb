```python
%matplotlib tk
import scipy as sc
import numpy as np
import sympy as sp
import matplotlib as mpl
import matplotlib.pyplot as plt
import scienceplots

plt.style.use(['science','notebook', 'grid'])
```

```python
def dphidt_f(phi_prime):
    return(phi_prime)

def dphi_primedt_f(t, l, R, g, phi, phi_prime):
    return (g*np.cos(phi)-R*phi_prime**2)/(l-R*phi)
```

```python
def dSdt(S, t, l, R, g):
    phi, phi_prime = S
    return [
        dphidt_f(phi_prime),
        dphi_primedt_f(t, l, R, g, phi, phi_prime)
    ]
```

```python
N = 1000
t = np.linspace(0,10, N)
l = 10
R = 1
g = 9.8

ans = sc.integrate.odeint(dSdt, y0=[0.5, 0], t=t, args=(l, R, g))
```

```python
phi = ans.T[0]
plt.plot(t, phi-np.pi/2)
tau = 2*np.pi*np.sqrt((l-R*np.pi/2)/g)
plt.axvline(tau, linestyle='dashed', c='r')
```
